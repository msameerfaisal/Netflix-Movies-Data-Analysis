```python
q1 = df.dropna(subset=["date_added"]).copy()
```

```python
q1.isna().count()
```

```python
yearly_counts = q1["release_year"].value_counts()
print(yearly_counts)
```

```python
fig,axes = plt.subplots(figsize=(12,6))
axes.plot(yearly_counts.index,yearly_counts.values,color="red",linewidth=3)
axes.set_xlabel("Years")
axes.set_ylabel("Number Of Movies Produced")
axes.set_title("Number Of Movies Per Year")
fig.savefig("PLt")
```

```python
q1[q1["release_year"] <= 2008][["release_year", "date_added"]]
```

```python
q1["year_added"] = q1["date_added"].dt.year
```

```python
q1.head()
```

```python
year_counts = q1["year_added"].value_counts().sort_index(ascending=False)
print(year_counts)
```

```python
fig,axes = plt.subplots(figsize=(12,6))
axes.plot(year_counts.index,year_counts.values,color="red",linewidth=3)
axes.set_xlabel("Year")
axes.set_ylabel("Number Of Titles Added At Netflix")
axes.set_title("Number Of Titles Added At Netflix Per Year")
fig.savefig("NumberOfTitlesAddedAtNetflixPerYear.tiff")
```
