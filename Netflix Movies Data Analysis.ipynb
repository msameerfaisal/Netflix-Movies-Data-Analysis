```python
yearly_counts = q1["release_year"].value_counts()
print(yearly_counts)
```

```python
fig,axes = plt.subplots(figsize=(12,6))
axes.plot(yearly_counts.index,yearly_counts.values,color="red",linewidth=3)
axes.set_xlabel("Years")
axes.set_ylabel("Number Of Movies Produced")
axes.set_title("Number Of Movies Per Year")
fig.savefig("PLt")
```
